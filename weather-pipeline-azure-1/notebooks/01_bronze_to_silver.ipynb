```python
path = "abfss://bronze@strgproject.dfs.core.windows.net/weather_data"

df = spark.read.format("json").load(path)
display(df)
```

```python
from pyspark.sql.functions import explode, arrays_zip, col

h = df.select("hourly.*")

df_rows = (h.select(explode(arrays_zip(
                "time", "temperature_2m", "wind_speed_10m", "precipitation",
                "relative_humidity_2m", "apparent_temperature", "cloud_cover",
                "visibility", "soil_temperature_0cm")).alias("r"))
            .select("r.*")
            .withColumn("time", col("time").cast("timestamp")))

display(df_rows)
print(df_rows.count())
```

```python
df_units = (df_rows
    .withColumnRenamed("temperature_2m", "temperature_c")
    .withColumnRenamed("apparent_temperature", "apparent_temperature_c")
    .withColumnRenamed("soil_temperature_0cm", "soil_temperature_c")
    .withColumnRenamed("wind_speed_10m", "wind_speed_kmh")
    .withColumnRenamed("precipitation", "precipitation_mm")
    .withColumnRenamed("relative_humidity_2m", "humidity_pct")
    .withColumnRenamed("cloud_cover", "cloud_cover_pct")
    .withColumnRenamed("visibility", "visibility_m"))

display(df_units)
```

```python
from pyspark.sql.functions import col, to_date, hour, current_timestamp

df_cols = (df_units
    .withColumn("weather_date", to_date(col("time")))
    .withColumn("hour_of_day", hour(col("time")))
    .withColumn("ingestionTime", current_timestamp()))

display(df_cols)
```

```python
from pyspark.sql.functions import col, count, when

null_counts = df_cols.select(
    [count(when(col(c).isNull(), c)).alias(c) for c in df_cols.columns]
)
display(null_counts)
```

```python
df_clean = (df_cols
    .filter(col("humidity_pct").between(0, 100))
    .filter(col("cloud_cover_pct").between(0, 100))
    .filter(col("wind_speed_kmh") >= 0)
    .filter(col("precipitation_mm") >= 0)
    .filter(col("visibility_m") >= 0)
    .dropDuplicates(["time"]))

print("before:", df_cols.count())
print("after:", df_clean.count())
```

```python
silver_table = "asadproject.schemaproject.silver_weather"

(df_clean.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(silver_table))
```

```python
display(spark.table(silver_table))
print(spark.table(silver_table).count())
```

```python
silver_table = "asadproject.schemaproject.silver_weather"
silver_path = "abfss://silver@strgproject.dfs.core.windows.net/weather"

spark.sql(f"DROP TABLE IF EXISTS {silver_table}")

(df_clean.write
    .format("delta")
    .mode("overwrite")
    .option("path", silver_path)
    .saveAsTable(silver_table))
```
